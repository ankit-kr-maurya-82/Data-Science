```python
# calculator

a= int(input("Enter first number: "))
b= int(input("Enter first number: "))
addl= lambda: a+b
sub= lambda: a-b
mul= lambda: a*b
div= lambda: a/b

def cal():
    print(addl())
    print(sub())
    print(mul())
    print(div())

cal()
```

```python
 # recursion
def count(n):
  
    print( n,"hello")
    if n==0:
        return
    count(n-1)

count(4)
```

```python

def add(n):
    if n==0:
        return 0
    return n+add(n-1)
add(4)
```

```python

def fact(n):
    if n==0:
        return 0
    return n*fact(n-1)
fact(4)
```
