```python
# project day 1
# Marksheet System Project
# Project Objective
# Create a Python program that asks for a student's name and marks, calculates the total and percentage, and displays the result


print("========================================")
print("          MARKSHEET SYSTEM              ")
print("========================================")

name=input("Enter Student Name: ")
py=int(input("Enter Python Marks: "))
math=int(input("Enter Maths Marks: "))
eng=int(input("Enter English Marks: "))
total_marks=py+math+eng
percentage=total_marks/3

print("---------------RESULT---------------")
print("Student Name\t : ",name)
print("Total Marks\t : ",total_marks)
print("Percentage\t : ",percentage)
print("Thank You!")
print("========================================")
```
