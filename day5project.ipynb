```python
def line(n=45):
    print("=" * n)

def add(a, b):
    return a+b

def sub(a, b):
    return a-b

def mul(a, b):
    return a*b

def div(a, b):
    if b == 0:
        return "Error (Division by zero)"
    return a/b

def cube(a):
  return a**3

def square(a):
  return a**2

def factorial(n):
  if n==0:
    return
  return n*factorial(n-1)


def output(a, b):
    line(15)
    print(f"Inputs:\nValue A: {a}\nValue B: {b}")
    line(15)

def show_calculate():
    line(30)
    print("\t\tCHOOSE CATEGORY")
    line(30)
    print("1. Basic Calculator")
    print("2. Algebra and Maths")
    line(30)
    return input("Select category (1 or 2): ").strip()

def main():
    category = show_calculate()

    if category == "1":
        basic_cal=["Addition", "Substraction", "Multiplication", "Division"]
        for i in range(0, len(basic_cal)):
          print(f"{i+1}. {basic_cal[i]}")
        operator = int(input("Enter choice: ").strip())
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))

        output(a, b)

        if operator==1:
            print("Result: ", add(a, b))
        elif operator==2:
            print("Result: ", sub(a, b))
        elif operator==3:
            print("Result: ", mul(a, b))
        elif operator==4:
            print("Result: ", div(a, b))
        else:
            print("Invalid Operator selected.")

    elif category == "2":
        print("\n[Algebra and Maths Module]")
        algebra_cal=["Cube", "Square", "Factorial"]

        for i in range(0, len(algebra_cal)):
          print(f"{i+1}. {algebra_cal[i]}")

        choice=int(input("Enter a choice: "))
        a = int(input("Enter  number: "))

        if choice==1:
          print("Result: ", cube(a))
        elif choice==2:
          print("Result: ", square(a))
        elif choice==3:
          print("Result: ", factorial(a))
    else:
        print("Invalid category selection.")

main()
```
