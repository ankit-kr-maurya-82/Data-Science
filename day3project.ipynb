```python

print("="*45)
print("         CRICKET SCORE ANALYZER")
print("="*45)

overs = int(input("Enter Number of Overs Played : "))

total_runs = 0
fours = 0
sixes = 0
dot_balls = 0

for over in range(1,overs+1):

    print(f"\n---------- Over {over} ----------")

    for ball in range(1,7):

        runs = int(input(f"Ball {ball} Runs (0-6) : "))
        print(f"Ball : {ball} Runs : {runs}")

        
        if runs<0 or runs>6:
            print("Invalid Runs! Counting as Dot Ball.")
            runs=0

        total_runs += runs

        if runs==4:
            fours += 1
        elif runs==6:
            sixes += 1
        elif runs==0:
            dot_balls += 1

total_balls = overs*6
boundary_runs = (fours*4)+(sixes*6)
strike_rate = (total_runs/total_balls)*100

if strike_rate>=150:
    performance="Excellent"

elif strike_rate>=100:
    performance="Good"

else:
    performance="Needs Improvement"

print("\n"+"="*45)
print("               SCORECARD")
print("="*45)

print(f"Total Overs     : {overs}")
print(f"Total Balls     : {total_balls}")
print(f"Total Runs      : {total_runs}")

print("-"*45)

print(f"Fours Hit       : {fours}")
print(f"Sixes Hit       : {sixes}")
print(f"Dot Balls       : {dot_balls}")

print("-"*45)

print(f"Boundary Runs   : {boundary_runs}")
print(f"Strike Rate     : {round(strike_rate, 2)}")

print("-"*45)

print(f"Performance     : {performance}")

print("="*45)

if total_runs>=100:
    print("Century Partnership Level Score!")

elif total_runs>=50:
    print("Half-Century Level Score!")

else:
    print("Keep Practicing!")

print("="*45)
print("          THANK YOU FOR PLAYING")
print("="*45)
```
