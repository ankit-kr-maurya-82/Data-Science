```python
df={
    "number_courses": data["number_courses"],
    "time_study": data["time_study"],
    "Marks": data["Marks"]
}
```

```python
# Dataset
dt=md.pd.DataFrame(df)

x=dt[["number_courses","time_study"]]
y=dt["Marks"]

model=md.LinearRegression()

model.fit(x,y)

print("Coefficients:", model.coef_[0])
print("Intercept:", model.intercept_)

courses=int(input("Enter number of courses:"))
study=float(input("Enter time of study:"))

marks=model.predict([[courses, study]])

print(f"Predicted Marks = {marks[0]:.2f}")


# md.plt.style.use('default')
md.plt.style.use('dark_background')
md.plt.figure(figsize=(7,5))
md.plt.scatter(dt["number_courses"], df["Marks"])
md.salary_pred = model.predict(x)
md.plt.plot(df["Marks"])
md.plt.xlabel("number_courses ")
md.plt.ylabel("Marks")
md.plt.title("Marks Linear Regression")
md.plt.grid()
md.plt.legend()
md.plt.show()
```
