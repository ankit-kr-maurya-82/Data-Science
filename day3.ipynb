```python
# even number
i=1
n=int(input("Enter number: "))
while (i<=n):
  if i%2==0:
    print(i, end=" ")
  i+=1
```

```python
# odd number
i=1
n=int(input("Enter number: "))
while (i<=n):
  if i%2!=0:
    print(i, end=" ")
  i+=1
```
