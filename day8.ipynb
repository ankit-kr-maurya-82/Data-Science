```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 40, 10, 60, 50]
z = [25, 10, 20, 0, 80]
a = [4, 67, 45, 98, 45]
b = [76, 22, 77, 22, 55]
plt.plot(x,z, label="python")
plt.plot(x,y, label="java")
plt.plot(x,a, label="java")
plt.plot(x,b, label="java")
plt.grid(True)
plt.legend()

plt.show()
```

```python
students= ["A", "B", "C", "D"]
marks = [80,90,70, 75]
plt.bar(students, marks)
colors = ['black', 'darkgrey', 'lightgrey', 'grey']
plt.bar(students, marks, color=colors)
plt.grid(True)
plt.legend()
plt.show()
```
