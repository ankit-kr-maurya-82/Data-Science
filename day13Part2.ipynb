## COVID-19 checkup Model

```python
import module as md
```

```python
data={
    "Fever": [98,99,100,101,102,103,98,99,101,102],
    "Cough": [0,0,1,1,1,1,0,1,1,1],
    "BreathingProblem":[0,0,0,1,1,1,0,0,1,1],
    "COVID_Status":[
        "Negative", "Negative", "Negative",
        "Positive","Positive","Positive",
        "Negative", "Negative", "Positive","Positive"
    ]
}

df= md.pd.DataFrame(data)

x=df[["Fever", "Cough", "BreathingProblem"]]
y=df["COVID_Status"]

model= md.DecisionTreeClassifier()

model.fit(x,y)

fever=float(input("Enter Temperature (°F)): "))
cough=float(input("Cough?  (1=Yes, 0=No): "))
breathing=float(input("Breathing  Problem (1=Yes, 0=No): "))

prediction=model.predict([[fever, cough, breathing]])
print("COVID Status: ", prediction[0])

y_pred=model.predict(x)
accuracy=md.accuracy_score(y,y_pred)
print("Accuracy:", accuracy*100, "%")
```
