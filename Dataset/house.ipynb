```python
df=df[['bedrooms','bathrooms','sqft_living','floors','price']]
df
```

```python
df=df.drop_duplicates()
df
```

```python
print(df)
```

```python
print(df.duplicated().sum())
```

```python
x = df[["sqft_living"]]
y = df["price"]

model = LinearRegression()
model.fit(x, y)

print("slope", model.coef_)
print("intercept", model.intercept_)

try:
    living_space = int(input("Enter the sqft_living to predict price: "))
    predicted_price = model.predict([[living_space]])
    print(f"Predicted price for {living_space} sqft: ${predicted_price[0]:,.2f}")
except ValueError:
    print("Please enter a valid number for sqft_living.")

plt.figure(figsize=(10, 6))
plt.scatter(x, y, color="olive", alpha=0.3, label="Actual Data")
plt.plot(x, model.predict(x), color="grey", linewidth=2, label="Regression Line")
plt.xlabel("Sqft Living")
plt.ylabel("Price")
plt.title("House Price vs Sqft Living")
plt.legend()
plt.show()
```

```python
x = df[["bedrooms"]]
y = df["price"]

model = LinearRegression()
model.fit(x, y)

print("slope", model.coef_)
print("intercept", model.intercept_)

try:
    living_space = int(input("Enter the sqft_living to predict price: "))
    predicted_price = model.predict([[living_space]])
    print(f"Predicted price for {living_space} sqft: ${predicted_price[0]:,.2f}")
except ValueError:
    print("Please enter a valid number for bedrooms")

plt.figure(figsize=(10, 6))
plt.scatter(x, y, color="yellow", alpha=0.3, label="Actual Data")
plt.plot(x, model.predict(x), color="black", linewidth=2, label="Regression Line")
plt.xlabel("bedrooms")
plt.ylabel("Price")
plt.title("House Price vs bedrooms")
plt.legend()
plt.show()
```
